# 1.1 Python小游戏

```python
import random
#游戏初始化
number = random.randint(1, 100) #生成1到100的随机数
attempts = 0
print("欢迎来到猜数字游戏！")
print("我想了一个1到100之间的数字，你能猜中吗？")
#游戏主循环
while True:
    guess = int(input("请输入你的猜测："))
    attempts += 1
    if guess < number:
        print("太小了！再试一次。")
    elif guess > number:
        print("太大了！再试一次。")
    else:
        print(f"恭喜你猜中了！这个数字是{number}。你总共猜了{attempts}次。")
        break
```

## 1.2.1导入模块和随机数生成

```python
import random
#random.randrange(stop)生成一个从0开始，到stop-1结束的随机整数。
random_number = random.randrange(10)
print(f"随机生成的数字是： {random_number}")
#random.randint(start,stop)生成一个范围在 [a, b] 之间的随机整数
random_number = random.randint(0,9)
print(f"随机生成的数字是： {random_number}")
#random.choice(range(stop)),choice() 函数的作用是从一个序列中随机抽取一个元素。 [a, b)
random_number = random.choice(range(10))
print(f"随机生成的数字是： {random_number}")
#random()生成一个[0.0,1.0)的小数
random_number = random.random()
print(f"随机生成的数字是： {random_number}")

```

## 1.2.3打印输出和字符串格式化

```python
#1.普通打印
print("Hi, welcome to the game!")
#2.打印变量，使用逗号连接字符串和变量，逗号会输出为空格
name = "Alice"
print("Welcome", name, "to the game!")
#3.使用f-string格式化输出(推荐)
guess_count = 3
print(f"Hi {name}, you have {guess_count} chances to guess the number.")
# 在f-string方法下设置小数位，居中和总宽度
pi = 3.141592653589793
print(f"The value of pi is approximately {pi:.2f}") #限制小数点后两位
print(f"|{'Hello ':^10}|") #^字符串居中对齐，总宽度为10
#4. 使用format方法并设置小数位
print("Hi {}, you have {} chances to guess the number.".format(name,
guess_count))
print("The value of pi is approximately {:.3f}".format(pi))
```

在 f-string 中，{} 里面不光能放“提前定义好的变量名”，它其实是一个表达式求值区。
核心规则是：只要是一小段“能算出具体结果的 Python 代码（表达式）”，都可以直接塞进 {} 里，根本不需要提前定义！最后变成文字

```python
print(f"我想直接打印数字：{999}") 
print(f"我想直接打印文字：{'Apple'}")
print(f"一天有 {24 * 60 * 60} 秒。")
# 输出：一天有 86400 秒。
print(f"alice变大写是：{'alice'.upper()}") 
# 输出：alice变大写是：ALICE

print(f"10和20哪个大？答案是：{max(10, 20)}")
# 输出：10和20哪个大？答案是：20
age = 15
print(f"你是个{'成年人' if age >= 18 else '未成年人'}。")
# 输出：你是个未成年人。
```

## 1.3.1简单综合练习：一年级小朋友的数学测试
  编写一个简单的数学测验游戏，使用 Python 语言。程序应包含以下功能：

  程序启动时，欢迎用户并说明游戏规则：用户需要回答 5 道随机生成的加法问题，每题涉及两个 0 到 20 之间的整数。问题生成与回答：程序随机生成两个 0 到 20 之间的整数，并显示给用户。用户输入这两个数的和作为答案。程序判断用户的答案是否正确：如果正确，打印“回答正确！”，并记录一次得分。如果错误，打印“回答错误！正确答案是 X”。游戏结束与得分显示：在用户回答完 5 道问题后，程序显示用户的总得分。
  
  也请适当改进游戏，让它更好玩吧。

```python
print('欢迎！你需要回答5道20以内加法题，对一题就得一分')
import random
count = 0
marks = 0
while count < 5:
    count +=1
    a = random.randint(0,20)
    b = random.randint(0,20)
    print(f'{a}+{b}=?')
    user_answer = int(input())
    if a + b == user_answer:
        print("正确")
        marks += 1
    else:
        print(f"回答错误！正确答案是{a + b}。")
print(f"总得分{marks}")
```

## 1.3.2复杂综合练习：MBTI 测试
请编写一个简单的 MBTI 测验游戏，使用 Python 语言。程序启动时，欢迎用户，简单说明规则：用户要回答 12 道随机生成的题目。程序会根据用户的回答计分。游戏结束后，告诉用户他的性格类型。

```python
import random
# 列表里包含了20个子列表，每个子列表格式为：["题目", "选项A", "选项B", "A对应的类型", "B对应的类型"]
questions = [
    ["在社交场合中，你更倾向于：", "A. 和很多人交流，充满活力", "B. 与少数人深度交谈，节省能量", "E", "I"],["当你有空闲时间时，你更喜欢：", "A. 外出参加活动或聚会", "B. 独自在家放松或做个人感兴趣的事情", "E", "I"],["在团队合作中，你更倾向于：", "A. 主动提出想法并快速行动", "B. 观察、思考后再表达自己的看法", "E", "I"],["与陌生人互动时，你通常：", "A. 感到自然并享受与他们的交流", "B. 感到不自在，更愿意保持距离", "E", "I"],
    ["在学习新事物时，你更关注：", "A. 具体的事实和实际应用", "B. 背后的概念和可能性", "S", "N"],["当描述一个想法时，你更倾向于：", "A. 使用具体的例子和细节", "B. 强调整体的愿景和理论", "S", "N"],["在做决策时，你更依赖：", "A. 直观的经验和实际数据", "B. 未来的潜力和可能的变化", "S", "N"],["你更擅长：", "A. 关注眼前的实际情况并解决问题", "B. 想象可能的情况并提出创新的解决方案", "S", "N"],["你更喜欢：", "A. 具体的、真实的材料", "B. 抽象的、理论性的内容", "S", "N"],
    ["当处理冲突时，你更倾向于：", "A. 逻辑地分析问题并寻找合理的解决方案", "B. 考虑他人的感受并尽量避免伤害", "T", "F"],["在做决定时，你更倾向于：", "A. 以客观事实为依据", "B. 以个人价值观为依据", "T", "F"],["你认为优秀的领导者应该：", "A. 冷静、理性、以目标为导向", "B. 关怀他人、以人为本", "T", "F"],["当朋友向你倾诉烦恼时，你更倾向于：", "A. 提供具体的建议或解决方案", "B. 表示理解并给予情感支持", "T", "F"],["当面对复杂问题时，你更倾向于：", "A. 分析问题的逻辑并寻找最佳方案", "B. 考虑问题对他人和团队的影响", "T", "F"],
    ["你更喜欢日常生活：", "A. 有计划并按部就班地进行", "B. 灵活自由，随时调整安排", "J", "P"],["当面对一个任务时，你更倾向于：", "A. 提前规划并按计划执行", "B. 临近截止日期时才集中精力完成", "J", "P"],["在制定长期目标时，你更倾向于：", "A. 具体化目标并按步骤实现", "B. 设定大致方向并适应变化", "J", "P"],["在家居环境中，你更喜欢：", "A. 整洁有序的空间", "B. 随意舒适的摆放", "J", "P"],["当计划被打乱时，你通常：", "A. 感到不安并尝试重新规划", "B. 自然接受并寻找替代的解决方法", "J", "P"],["你更倾向于：", "A. 事先明确任务细节并准备好", "B. 在行动中逐步调整计划", "J", "P"]
]
print("欢迎来到 MBTI 性格测试游戏！")
print("规则：你将回答 12 道随机生成的题目。")
print("请根据你的第一直觉，输入 A 或 B。")
selected_questions = random.sample(questions, 12)
scores = {"E": 0, "I": 0, "S": 0, "N": 0, "T": 0, "F": 0, "J": 0, "P": 0}
for count in range(12):
    current_q = selected_questions[count] 
    print(f"第 {count + 1} 题：{current_q[0]}")
    print(current_q[1])
    print(current_q[2])
    answer = input("请输入你的选择（A 或 B）：").upper()
    if answer == "A":
        scores[current_q[3]] += 1
    elif answer == "B":
        scores[current_q[4]] += 1
    else:
        print("输入无效，本题作废不计分。")
type_result = ""
if scores["E"] >= scores["I"]:
    type_result += "E"
else:
    type_result += "I"
if scores["S"] >= scores["N"]:
    type_result += "S"
else:
    type_result += "N"
if scores["T"] >= scores["F"]:
    type_result += "T"
else:
    type_result += "F"
if scores["J"] >= scores["P"]:
    type_result += "J"
else:
    type_result += "P"
print(f"你的 MBTI 性格类型是：{type_result}")
```
