## 2.1.1字典
键必须唯一，类似单词；值可以重复，类似解释。

```python
# 定义一个字典：存储单词和它的解释
dictionary = {
"apple": "一种水果，红色或绿色，味道甜",
"python": "一种编程语言，简单易学，功能强大",
"table": "一种家具，通常用来放置物品"
}
# 查询单词的解释
print(dictionary["apple"]) 
# 添加新单词
dictionary["book"] = "用纸张装订成册的阅读材料"
print(dictionary)
# 修改单词的解释
dictionary["python"] = "一种蛇，也是一种编程语言"
print(dictionary)
# 删除单词
del dictionary["table"]
print(dictionary)
```

## 2.1.2列表
列表可以存储任意类型的元素，比如字符串、数字，甚至是另一个列表或字典。

```python
# 定义一个字典：每个维度对应的问题列表
questions = {
"EI": [
("你更喜欢：","A.热闹的聚会","B.安静的独处"),
("你更倾向于：","A.说出你的想法","B.先思考再说")
],
"SN": [
("你更关注：", "A.现实中的细节","B.长远的可能性"),
("在做决定时，你更依赖：","A.具体事实","B.抽象概念")
]
}
# 查字典：查询EI键所对应的值
print(questions["EI"])
# 查列表：遍历列表-SN键对应的值中的每个元素
for j in questions["SN"]:
    print(j)
# 输出每一个问题的A选项
for i in questions:
    for j in questions[i]:
        print(j[1])
```
