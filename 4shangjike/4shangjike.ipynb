## 4.1知识引入
- 等额本息：每月还款的总额（本金 + 利息）是固定的。
    - 优点： 每月支出固定，便于个人或企业做财务预算。
    - 缺点： 相比等额本金，总利息支出较高，因为前期本金偿还速度较慢。
- 等额本金：每月偿还的本金是固定的，利息则根据剩余本金逐月计算。
    - 优点： 总利息支出较少。由于前期偿还本金快，欠款基数下降迅速，产生的利息随之快速减少。
    - 缺点： 前期还款压力巨大，对借款人的即期现金流要求较高。

以下是等额本息下贷款期限从 3 年到 70 年的每月还款额变化趋势

````python
import os 
import matplotlib.pyplot as plt
# 若目录不存在则创建 
os.makedirs(r"D:\jjdsjyrgzn_python\4shangjike", exist_ok=True) 
# 切换到工作目录
os.chdir(r"D:\jjdsjyrgzn_python\4shangjike") 
# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei'] 
def loan_calc(principal , annual_rate , years):
    #计算月利率
    r = annual_rate / 12 
    # 总期数（月）
    n = years * 12 
    # 等额本息公式：每月还款额 = P×(r×(1+r)^n) / ((1+r)^n−1)
    return principal * (r * (1 + r)**n) / ((1 + r)**n - 1)
# 贷款金额（80万元）
p = 800000 
# 年利率（5%）
r = 0.05 
# 贷款年限从3~70年
yr = range(3, 71) 
# 计算每种年限下的每月还款额
mp = [loan_calc(p, r, y) for y in yr]
# 绘图
# 蓝色折线
plt.plot(yr, mp, 'b-o', label='每 月 还 款 额') 
plt.xlabel('贷 款 年 限 （年） ') 
plt.ylabel('每 月 还 款 额 （元） ') 
# 保存图表到工作目录
plt.savefig('monthly_payments.png', dpi=100) 
plt.show() 
````

## 4.3综合练习
### 简单综合练习
计算贷款本金为 80 万元，年利率为 4.5%，贷款期限为 3-70年时的总还款额与总利息额，并画图表示。


````python
import matplotlib.pyplot as plt
# 1. 绘图字体与负号设置
plt.rcParams['font.sans-serif'] = ['SimHei'] 
plt.rcParams['axes.unicode_minus'] = False 
# 2. 定义计算函数（返回三个值）
def loan_calc_all(principal, annual_rate, years):
    """返回每月还款额、总还款额、总利息额"""
    r = annual_rate / 12           # 月利率
    n = years * 12                 # 总期数（月）
    # 计算每月还款额
    monthly_payment = principal * (r * (1 + r)**n) / ((1 + r)**n - 1)
    # 计算总还款额 = 月供 × 总月数
    total_payment = monthly_payment * n
    # 计算总利息额 = 总还款额 - 贷款本金
    total_interest = total_payment - principal
    return monthly_payment, total_payment, total_interest
# 3. 设置参数
p = 800000       
rate = 0.045     
yr = range(3, 71)
# 4. 批量计算
# 先把所有年限的结果计算出来存成一个大列表
results =[loan_calc_all(p, rate, y) for y in yr]
# 用列表推导式将结果拆分（res[0]是月供，res[1]是总还款，res[2]是总利息）
print(results)
t_payments = [res[1] for res in results]
t_interests = [res[2] for res in results]
# 5. 绘图
plt.figure(figsize=(10, 6)) #
plt.plot(yr, t_payments, 'r-', linewidth=2, label='总还款额 (元)')
plt.plot(yr, t_interests, 'g-', linewidth=2, label='总利息额 (元)')
# 画一根参考线：贷款本金（常数80万）
plt.axhline(y=p, color='b', linestyle='--', alpha=0.6, label='初始贷款本金 (80万元)')
plt.xlabel('贷款年限 (年)')
plt.ylabel('金额 (元)')
plt.title('贷款年限变化对总还款额与总利息额的影响\n(本金:80万, 利率:4.5%)')
plt.legend()       
plt.grid(True, linestyle=':', alpha=0.7) 
plt.savefig('total_payment_interest.png', dpi=150)
plt.show()
````

题目要求思考： 贷款年限对总利息额的影响大吗？为什么？

解答：影响非常大！并且这种影响是非线性的。图表会清楚地显示：随着年限拉长，绿色的“总利息额”曲线呈近乎上升。当贷款达到约 33 年时，总利息额就会超过 80 万（利息 > 本金）；如果拉长到 70 年，总利息更是会达到本金的好几倍。
### 复杂综合练习 - 为不同收入情况的人分析房贷方案
假设有以下四类收入情况的人群：
- 月入 5000 人群：月收入 5000 元。
- 月入 1 万人群：月收入 10,000 元
- 月入 2 万人群：月收入 20,000 元。
- 月入 5 万人群：月收入 50,000 元。

每个人希望根据自己的收入选择合适的房贷方案，满足以下条件，为了简化分析，我们假定每个人都要贷款 100 万，此时你需要为每个人选择可行的、尽可能优化的房贷期限：

- 每月还款额不超过月收入的 50%。
- 房贷方案包括不同的贷款金额（principal）和贷款年限（years），年利率固定为4.5%。

练习要求：
- 完成代码，根据收入情况分析不同方案的可行性。
- 测试不同的贷款金额和年限对方案可行性的影响。
- 思考：月入 5000 和月入 50000 人群选择房贷方案时，可能会有什么不同的策略？如果年利率上浮到 5%，房贷方案的可行性会受到怎样的影响？对于收入较低的人群，延长贷款年限是否是一个好的策略？为什么？

````python
import matplotlib.pyplot as plt
# 1. 绘图字体与负号设置
plt.rcParams['font.sans-serif'] = ['SimHei'] 
plt.rcParams['axes.unicode_minus'] = False 
# 2. 定义函数
def loan_calc(principal, annual_rate, years):
    """计算等额本息每月还款额"""
    r = annual_rate / 12 
    n = years * 12 
    return principal * (r * (1 + r)**n) / ((1 + r)**n - 1)
# 3. 设定统一参数
principal = 1000000    
rate = 0.045           
years_list = range(5, 55, 5) 
# 4. 定义四类人群的月收入
income_groups =[
    {"title": "月入5000", "income": 5000},
    {"title": "月入1万", "income": 10000},
    {"title": "月入2万", "income": 20000},
    {"title": "月入5万", "income": 50000}
]
# 5. 绘制 2x2 的组合子图
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
# 添加大标题
fig.suptitle('不同收入人群的房贷可行性（贷款100万，年利率4.5%，最长50年）', fontsize=16)
# 将 2x2 的矩阵打平为一维，方便在循环中遍历
axes = axes.flatten()
for i, group in enumerate(income_groups):
    ax = axes[i]
    income = group["income"]
    # 监管/风控红线：月供不得超过月收入的 50%
    max_payment = income * 0.5 
    payments = []
    colors =[] # 用于区分可行(蓝)和不可行(红) 
    # 计算不同年限下的情况
    for y in years_list:
        mp = loan_calc(principal, rate, y)
        payments.append(mp)
        if mp > max_payment:
            colors.append('#E26A6A') # 红色代表不可行 (超过50%红线)
        else:
            colors.append('#5B9BD5') # 蓝色代表可行        
    # X轴刻度转换为字符串 "5年", "10年"
    x_labels =[f"{y}年" for y in years_list]  
    # 绘制柱状图
    ax.bar(x_labels, payments, color=colors)  
    # 画一条水平虚线代表“还款能力上限”
    ax.axhline(max_payment, color='gray', linestyle='--', linewidth=1.5, 
               label=f'50%收入红线 ({int(max_payment)}元)') 
    # 子图装饰
    ax.set_title(group["title"])
    ax.set_ylabel('每月还款额 (元)')
    ax.tick_params(axis='x', rotation=45)
    # 固定Y轴的范围，让四张图比例一致，更具直观对比性
    ax.set_ylim(0, 21000) 
    ax.legend()
# 自动调整子图间距防重叠
plt.tight_layout(rect=[0, 0, 1, 0.96]) 
plt.savefig('feasibility_analysis.png', dpi=150)
plt.show()
````

思考 1：月入 5000 和月入 50000 人群选择房贷方案时，可能会有什么不同的策略？

- 月入 5000（底层杠杆受限）：
    - 现象： 看图可知，图1全是红柱。因为 100 万按 4.5% 哪怕贷 50 年，最低月供也要 4182 元，远超 2500 元的 50% 红线。
    - 策略： 他们根本无法通过时间来换取空间。必须改变核心参数：要么增加首付比例（降低 Principal）、要么买更便宜的房子、要么寻找共同借款人（合并夫妻收入来提高红线）。
- 月入 5万（高净值人群的财务优化）：
    - 现象： 看图可知，图4全是蓝柱，即使是最短的 5 年期月供（1.8万左右）也在 2.5万的红线内。所有方案均可行。
    - 策略： 他们的策略取决于机会成本。如果他们能找到年化收益率 >4.5%的理财/投资渠道，他们反而会主动选择 30 年长贷，把钱留在手里套利（借低买高）；如果他们极端厌恶风险，则可以选择 5-10 年极速还清，节省总利息。

思考 2：如果年利率上浮到 5%，房贷方案的可行性会受到怎样的影响？
- 影响：
    - 从微观上看，所有柱子的绝对高度都会上升。这意味着原本处于“及格线边缘”的方案会变为不可行（由蓝变红）。例如，月入1万的人原本贷 25 年恰好可行，利率上浮后可能被逼退到必须贷 30 年以上。
    - 从宏观经济调控（货币政策）来看，央行加息会直接“挤出”边缘购房者，这正是房地产去杠杆、压抑过热需求的经典手段。

思考 3：对于收入较低的人群，延长贷款年限是否是一个好的策略？为什么？
- 这并非是一个“好策略”，而是一种“被迫妥协”。
- 原因 1：边际效用递减的陷阱。随着年限从 30 年拉长到 50 年，每月的月供下降得越来越少（图中曲线越往后越平缓），因为这部分月供里几乎全是利息，本金摊销微乎其微。为了每个月少还一两百块钱，要多付出几十万的利息成本。
- 原因 2：跨期脆弱性。
低收入人群通常面临更高的失业和收入波动风险。拉长到 40-50 年意味着终生背负债务。生命周期中一旦遇到重病、经济下行等外部冲击，其脆弱的现金流极易断裂，导致抵押物被银行法拍。
