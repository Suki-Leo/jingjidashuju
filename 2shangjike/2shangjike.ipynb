## 2.1.1字典
键必须唯一，类似单词；值可以重复，类似解释。

```python
# 定义一个字典：存储单词和它的解释
dictionary = {
"apple": "一种水果，红色或绿色，味道甜",
"python": "一种编程语言，简单易学，功能强大",
"table": "一种家具，通常用来放置物品"
}
# 查询单词的解释
print(dictionary["apple"]) 
# 添加新单词
dictionary["book"] = "用纸张装订成册的阅读材料"
print(dictionary)
# 修改单词的解释
dictionary["python"] = "一种蛇，也是一种编程语言"
print(dictionary)
# 删除单词
del dictionary["table"]
print(dictionary)
```

利用字典相加字符串

```python
# 示例：根据分数拼接MBTI类型
scores = {"E": 2, "I": 1, "S": 2, "N": 3}
result = ""
result += "E" if scores["E"] > scores["I"] else "I"
result += "S" if scores["S"] > scores["N"] else "N"
print(f"你的MBTI类型是：{result}")
```

## 2.1.2列表
列表可以存储任意类型的元素，比如字符串、数字，甚至是另一个列表或字典。

```python
# 定义一个字典：每个维度对应的问题列表
questions = {
"EI": [
("你更喜欢：","A.热闹的聚会","B.安静的独处"),
("你更倾向于：","A.说出你的想法","B.先思考再说")
],
"SN": [
("你更关注：", "A.现实中的细节","B.长远的可能性"),
("在做决定时，你更依赖：","A.具体事实","B.抽象概念")
]
}
# 查字典：查询EI键所对应的值
print(questions["EI"])
# 查列表：遍历列表-SN键对应的值中的每个元素
for j in questions["SN"]:
    print(j)
# 输出每一个问题的A选项
for i in questions:
    for j in questions[i]:
        print(j[1])
```

## 2.1.3元组
元组中的数据一旦定义，就不能修改（不可增删改）。
元组适合用来存储固定的数据，比如地理坐标、RGB 颜色值等

```python
# 定义一个元组：存储问题和选项
example = ("你更喜欢：","A.热闹的聚会","B.安静的独处")
print(example[0]) # 输出： 你更喜欢：
print(example[1]) # 输出： A.热闹的聚会
print(example[2]) # 输出： B.安静的独处
```

## 2.2综合练习
### 2.2.1简单
某顾客在肯德基点了一份订单，订单包含以下商品：  
• 甜筒：2 支，每支单价 ¥8  
• 汉堡：2 个，每个单价 ¥15  
• 薯条：1 份，每份单价 ¥10  
请按照以下要求完成程序设计：
1. 使用字典存储订单信息，键为商品名称，值为一个列表，列表包含商品的数量和单价。
2. 遍历字典，计算订单总价。
3. 使用 f-string 格式化输出结果。

```python
order = {
"甜 筒": [2, 8], 
"汉 堡": [2, 15],
"薯 条": [1, 10]
}
price = 0
for i in order:
    price += order[i][0] * order[i][1]
print(f"总价是：{price}元")
```
